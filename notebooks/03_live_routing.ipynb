# Live routing of a new query

This optional notebook is **not required for paper reproduction**. It obtains fresh semantic evidence from Laya or TypeSafe JEV, converts the result to the same 40 ProbabilityMass features used to fit the released router, and ranks the 20 candidate models.

- `backend='laya'` is the default. Install with `uv sync --extra live-laya --extra notebooks`.
- `backend='jev'` uses `TYPESAFE_API_KEY` from `.env` via `python-dotenv`.

````python
import warnings
warnings.filterwarnings("ignore", message="IProgress not found")
warnings.filterwarnings("ignore", message="laya: this checkpoint ships temperatures")

from pathlib import Path

ROOT = Path.cwd().resolve()
if not (ROOT / 'pyproject.toml').exists():
    ROOT = ROOT.parent

backend = 'laya'  # 'laya' or 'jev'
query = 'Implement a Python parser that preserves comments, rejects duplicate keys, and reports the exact malformed line.'
print(query)
````

### JEV credentials

For JEV only, copy `.env.example` to `.env` and set:

```dotenv
TYPESAFE_API_KEY=your_key_here
```

The `.env` file is ignored by Git. Laya does not require a TypeSafe key.

````python
from selmroute.live import route_live

result = route_live(
    query,
    backend=backend,
    root=ROOT,
    top_k=5,
     env_file=ROOT / ".env",
    device=None,  # or "cuda" / "cpu" for Laya
    verbose=True,
)

result["selected_model"]
````

````python
import pandas as pd

pd.DataFrame(result['ranking'])
````
